# Corporate Information Confidentiality and Security Check Manual

---

## Foreword

The majority of information leakage incidents within enterprises do not originate from sophisticated hacker attacks, but rather from everyday office negligence, environmental vulnerabilities, device-related risks, and non-compliant personnel behaviour.

Confidentiality is not a specialised undertaking; it is a matter of routine, detailed management. This manual is applicable to: government and enterprise entities, classified office environments, technology firms, project-oriented companies, and business negotiation meeting rooms. It may be printed directly, distributed to all personnel, and employed for daily inspections, quarterly self-assessments, and rectification record-keeping.

## Chapter 1: Self-Inspection of Office Personnel Behaviour (High-Frequency Sources of Leakage)

### 1.1 Management of Mobile Phones and Smart Devices

1. Mobile phones, Bluetooth earphones, smart watches, and recording wristbands are strictly prohibited from classified meetings, internal decision-making meetings, and business negotiations.
2. Office mobile phones must not be installed with recording, cloud backup, image recognition, or scanning applications of unknown origin.
3. Voice assistants, automatic recording, and contextual memory functions on mobile phones must not be activated at workstations or in meeting rooms.
4. Photographing, screenshotting, or screen-recording classified documents, screens, or whiteboard content is prohibited.
5. Employees' personal mobile phones must not be connected to the office intranet or classified Wi-Fi networks.

### 1.2 File Transfer and Storage Behaviour

1. Classified documents must not be uploaded to the external internet, stored in the cloud, transmitted via WeChat, or backed up to network drives.
2. The use of third-party scanning applications or image recognition tools for processing internal materials is prohibited.
3. Internal documents must not be privately shared on social media, industry groups, or social platforms.
4. Departing or reassigned personnel must complete document handover, computer clearance, and permission revocation.

### 1.3 Oral Confidentiality Standards

1. Project progress, technical parameters, quotation proposals, and client information must not be discussed in public areas.
2. Internal classified work must not be discussed in lifts, canteens, lobbies, taxis, or other public places.
3. Internal testing procedures, protection schemes, and confidentiality architectures must not be disclosed to personnel from external organisations.

## Chapter 2: Environmental Confidentiality Self-Inspection of Meeting Rooms and Negotiation Areas (Key High-Risk Zones)

### 2.1 Site Environment Inspection

1. Wall corners, socket panels, decorative objects, plants, and areas surrounding projection equipment must be free from modified or concealed surveillance devices.
2. The underside of meeting tables, ceiling panels, and air-conditioning vents must be free from unidentified devices.
3. A comprehensive environmental re-inspection must be conducted following temporary construction, maintenance, or cabling work.
4. External glass and floor-to-ceiling windows should not display internal screens or documents directly for prolonged periods.

### 2.2 Meeting Equipment Security

1. Wireless microphones, conference speakers, and screen-casting devices should be powered down when not in use.
2. Computers, USB drives, and recording devices brought by external personnel must not be connected to the meeting system without authorisation.
3. Following important meetings: screen-casting should be terminated, whiteboards cleared, computer screens switched off, and desktop documents removed.

### 2.3 Personnel Access Management

1. External visitors, construction workers, and tour groups must be accompanied at all times and restricted to designated areas.
2. External personnel must not be left unaccompanied in meeting rooms or core office areas.

## Chapter 3: Self-Inspection of Office Equipment Against Leakage (High-Risk Area of Electromagnetic Leakage)

### 3.1 Computer Terminal Self-Inspection

1. Remote desktop, sharing permissions, and automatic cloud synchronisation should be disabled on computers.
2. Macros must not be enabled and editing must not be permitted for unfamiliar emails or attachments.
3. Computers should be configured with automatic screen locking, compliant password complexity, and immediate locking upon departure.
4. Private USB drives and external hard drives must not be inserted into office computers without authorisation.

### 3.2 Printer and Photocopier Self-Inspection

1. Printers and photocopiers contain built-in hard drives; professional data destruction must be performed prior to disposal or external repair.
2. Classified documents must be collected immediately after printing and must not be left at the output tray.
3. Unattended batch printing of sensitive materials is prohibited.

### 3.3 Wireless Device Security Self-Inspection

1. Wireless mice, wireless keyboards, and wireless screen-casting are prohibited in classified office areas.
2. Bluetooth, Wi-Fi, and hotspot functions that are not in long-term use should be manually disabled.
3. Wired-only office equipment should be prioritised in core areas to eliminate electromagnetic radiation leakage.

## Chapter 4: Network and Permission Security Self-Inspection

### 4.1 Account and Permission Management

1. Regular clearance of departed employees' accounts, outsourced accounts, and temporary visitor accounts.
2. One person, one account; shared accounts and shared passwords are prohibited.
3. High-privilege accounts are designated for specific personnel only and must not be lent or temporarily borrowed.

### 4.2 Network Environment Standards

1. Internal and external networks must be strictly physically isolated, with no mixed use or cross-connection.
2. Office Wi-Fi passwords should be changed regularly, with weak passwords and anonymous access disabled.
3. Unauthorised routers, private hotspots, and self-installed network devices are prohibited.

## Chapter 5: Site Electromagnetic Security and Physical Protection Self-Inspection

### 5.1 Electromagnetic Radiation Protection

1. Regular electromagnetic signal testing should be conducted in areas where computers, monitors, and servers are concentrated.
2. Classified rooms should avoid being adjacent to corridors, external walls, or public walkways.
3. Signal shielding and filtering or noise-reduction measures may be implemented in important meeting rooms as required.

### 5.2 Physical Anti-Theft and Anti-Intrusion Measures

1. Filing cabinets, document cabinets, and safes must be locked when unattended.
2. Paper classified materials must be shredded prior to disposal and must not be discarded directly.
3. All office equipment should be powered down at night when the premises are unoccupied.

## Chapter 6: Routine Inspection Self-Check Lists

### Daily Self-Check Items

- No classified documents left at workstations; no exposed information on screens
- Meeting rooms fully cleared after meetings; whiteboards erased
- All unattended devices locked and powered down
- No private devices brought into classified areas

### Weekly Self-Check Items

- Clear idle accounts and review abnormal login records
- Check wireless device usage
- Inspect public areas for unfamiliar devices

### Monthly Self-Check Items

- Visual inspection of the entire office area
- Compliance check of printer and photocopier usage
- Review of all personnel's confidentiality behaviour

### Quarterly Special Self-Check Items

- Full-site electromagnetic environment testing
- Device data security inspection
- Verification of confidentiality policy implementation

## Chapter 7: Red Line Prohibitions (Mandatory for All Personnel)

1. Classified information must not be connected to the internet; internet-connected devices must not contain classified information.
2. Unfamiliar files must not be opened; unfamiliar links must not be clicked.
3. Classified matters must not be discussed in public areas; classified information must not be stored on private devices.
4. External personnel must not be left out of sight; unfamiliar devices must not be connected to the intranet.
5. Discarded materials must be shredded; unattended devices must be locked.

## Conclusion

Confidentiality and security admit no trivialities. All major leakage incidents begin with a single act of negligence, a single instance of false optimism, or a single poor habit. Routine self-inspection, standardised implementation, and detailed management constitute the core protective barrier for enterprise information security.
